```python
import re
import numpy as np

def feet_to_cm(value):
  #extract the numbers
  numbers = re.match(r"(\d+)'(\d+)\"", value)
  if numbers:
    feet = int(numbers.group(1))
    inches = int(numbers.group(2))
    cm = ((feet*12) + inches)* 2.54
    return cm
  else:
    return value
```

There are some values in lb, thus, we have to convert them to kg and then delete "kg". After that, we can then convert the data type to int

```python
def lb_and_kg_to_numerical(value):

  if "lbs" in value:
    number = value.replace("lbs", "")
    number = float(number)
    kgs_value = number * 0.453592
    return kgs_value
  elif "kg" in value:
    kg_mod_value = value.replace("kg", "")
    return kg_mod_value
  else:
    return value

```

Now lets check the "Value", "Wage", and "Release Clause" columns

```python
df[["Value", "Wage", "Release Clause"]].value_counts(ascending=True)
```

```python
def monetary_to_numerical(value):
  value = str(value)
  if "€" in value:
    value = value.replace("€", "")
  if "K" in value:
    value = value.replace("K", "")
    value = float(value)
    value = value * 1000
    return value
  elif "M" in value:
    value = value.replace("M", "")
    value = float(value)
    value = value * 1000000
    return value
  else:
    return value

```

```python
df[["Value", "Wage", "Release Clause"]] = df[["Value", "Wage", "Release Clause"]].map(monetary_to_numerical)
```

```python
df[["Value", "Wage", "Release Clause"]]
```

```python
df = df.rename(columns={"Value": "Value (In €)"})
df = df.rename(columns={"Wage": "Wage (In €)"})
df = df.rename(columns={"Release Clause": "Release Clause (In €)"})
```
